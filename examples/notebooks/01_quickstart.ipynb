# Eden Quickstart

Generate a Gymnasium-compatible reinforcement-learning environment from a
natural-language description. Streams progress events in real time.

```python
%pip install eden-py
```

```python
import os
from eden import EdenClient

client = EdenClient(api_key=os.environ["EDEN_API_KEY"])

async def main():
    # Submit a job describing the environment in plain English
    job = await client.jobs.create(
        description="A 2D grid-world where the agent must collect keys to unlock a door before reaching a goal.",
        framework="gymnasium",
        tags=["rl", "gridworld"],
    )
    print(f"Submitted job: {job.id}")

    # Stream progress events as the pipeline runs
    async for event in client.jobs.stream(job.id):
        print(f"[{event.stage}] {event.message}")
        if event.type == "complete":
            return event.result

result = await main()
print(f"\nDone! Tier: {result.outcome_tier}")
```

```python
# Download the generated environment code as a ZIP
env_code = await client.environments.download(result.environment_id)
print(env_code[:500])
```
