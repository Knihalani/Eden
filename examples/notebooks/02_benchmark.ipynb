# Running a Benchmark

Use the parallel benchmark runner to evaluate a trained policy against a
versioned environment across multiple seeds.

```python
from eden import EdenClient
from eden.batch import ParallelJobRunner

client = EdenClient()
runner = ParallelJobRunner(client, max_concurrent=5)

# Run a benchmark across 20 seeds, in parallel
summary = await runner.run_benchmarks(
    environment_id="env_gridworld_v3",
    seeds=list(range(20)),
    policy="random",
    episodes_per_seed=10,
)

print(f"Mean reward: {summary.mean_reward:.2f} ± {summary.std_reward:.2f}")
print(f"Pass rate:   {summary.pass_rate * 100:.1f}%")
print(f"P95 latency: {summary.p95_latency_ms:.0f} ms")
```

```python
# Export the benchmark results as a dataset for fine-tuning
dataset = await runner.create_datasets_from_traces(
    summary.trace_ids,
    dataset_type="regression_suite",
)
print(f"Dataset: {dataset.id}  ({len(dataset.entries)} entries)")
```
